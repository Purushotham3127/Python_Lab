```python
#palindrome or not using function
def pal(n):
  x=0
  a=n
  while(a!=0):
    x=x*10+a%10
    a=a//10
  if(x == n):
    print("Given number is palinrome")
  else:
    print("Given number is not palindrome")
n=int(input("Enter the number : "))
pal(n)
```

```python
#reverse number using function
def rev(n):
  x=0
  a=n
  while(a!=0):
    x=x*10+a%10
    a=a//10
  return x
n=int(input("Enter the number : "))
i=rev(n)
print(i)
```

```python
#reverse a string
def reverse_str(s):
  rev=s[::-1]
  return rev
s=input()
a=reverse_str(s)
print(a)
```

```python
#multiple return values
def mul_val(*num):
    s = sum(num)
    l = len(num)
    avg = s / l
    return s, l, avg
r_s, r_l, r_a = mul_val(1, 2, 3)
print(f"sum {r_s} length {r_l} avg {r_a}")

```
