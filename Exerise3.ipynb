```python
#print odd numbers up to n
n = int(input())
for i in range(1,n+1,2):
  print(i,end=" ")
```

```python
#Print multiplication tabel of give number
n=int(input())
for i in range(1,11):
  print(f"{n} * {i} = {n*i}")
```

```python
#checking prime number
n = int(input())
for i in range(2,int(n/2+1)):
  if(n%i==0):
    print(f"{n} is not prime number")
    break
else:
  print(f"{n} is prime number")
```

```python
#print prime numbers betwn 2 numbers
n1 = int(input())
n2 = int(input())
for i in range(n1,n2):
  for j in range(2,i):
    if(i%j==0):
      break
  else:
    print(i,end = " ")
```

```python
#print the sum of digits in a given number
n = int(input())
sum=0
while(n!=0):
  sum=sum+n%10
  n=int(n/10)
print(sum)
```

```python
#reverse of given number
n=int(input("Enter the number :"))
sum=0
r=0
while n>0 :
    r=n%10
    sum=sum*10+r
    n=n//10

print(f"sum of digits of given number is : {sum}")
```
