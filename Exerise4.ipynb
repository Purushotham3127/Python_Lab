```python
#Lists in Python
L=[2,3,4,5,6,7]
print(sum(L))
print(max(L))
print(min(L))
print(len(L))
M= [4,8,9]
print(L+M)#concatination means add but does not change the list
print(L*3)#repetation
print(M[0])
print(8 in M)#membership operator
print(1 in M)
print(8 not in M)#membership operator
print(1 not in M)
#Iterating using a list
for i in L:
  print(i)
#Slicing
print(L[2:6:1])
print(L[::-1])#Used for reverse printing
print(L[0:5:2])
print(L[-1:-5:-1])#Lat element is indexed as -1 then -2 and so on
```
